```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Load the excel file

```python
file_path = "dataset/WalmartSales.xlsx"
```

```python
df = pd.read_excel(file_path)
```

## Display  the first few rows of the dataset

```python
df.head()
```

## A. Calculate the total sales revenue for each transaction

```python
df['Total Revenue'] = df['Unit price'] * df['Quantity']
```

## Group by City and Branch to analyze performance

```python
city_branch_performance = df.groupby(['City','Branch']).agg({
    'Total Revenue': ['sum', 'mean', 'count'],
    'Invoice ID': 'nunique'
}).reset_index()
```

## Rename columns for better readability

```python
city_branch_performance.columns = ['City', 'Branch', 'Total Sales Revenue', 'Average Sales Revenue', 'Total Transactions', 'Unique Invoices']
```

## Display the city and branch performance

```python
city_branch_performance
```

## B. Calculate the average price of an item sold aat each branch in each city

```python
avg_price_per_branch = df.groupby(['City','Branch'])['Unit price'].mean().reset_index()
```

```python
avg_price_per_branch.columns = ['City', 'Branch', 'Average Unit price']
```

```python
avg_price_per_branch
```

## C. Convert Date to datetime format

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df['Month'] = df['Date'].dt.month
```

```python
df['Year'] = df['Date'].dt.year
```

## Calculate the total sales revenue for each transaction

```python
df['Total Revenue'] = df['Unit price'] * df['Quantity']
```

```python
monthly_performance = df.groupby(['Year', 'Month', 'Product line', 'Gender', 'Payment']).agg({
    'Total Revenue': ['sum', 'mean', 'count']
}).reset_index()
```

## Rename columns

```python
monthly_performance.columns = ['Year', 'Month', 'Product Line', 'Gender', 'Payment Method', 'Total Sales Revenue', 'Average Sales Reveue', 'Total Transactions']
```

```python
april_2019_data = monthly_performance[(monthly_performance['Year'] == 2019)]
```
